# Join Customer and Address
Join customer data with address data to create a customer_address table which contains the address of each customer on the same record

```python
silver_customers = spark.table("gizmobox.silver.customers")
display(silver_customers)
```

```python
silver_addresses = spark.table('gizmobox.silver.addresses')
display(silver_addresses)
```

```python
df_golden = (
    silver_customers
    .join(silver_addresses, "customer_id", "inner")
    .select(
        "customer_id",
        "customer_name",
        "date_of_birth",
        "email",
        "member_since",
        "telephone",
        "billing_address_line_1",
        "billing_city",
        "billing_state",
        "billing_postcode",
        "shipping_address_line_1",
        "shipping_city",
        "shipping_state",
        "shipping_postcode"
    )
)
display(df_golden)
```

```python
df_golden.writeTo("gizmobox.gold.customer_address")
```

```python
df = spark.table("gizmobox.gold.customer_address")
display(df)
```
