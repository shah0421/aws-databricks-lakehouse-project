# Transform Memberships Data
1. Extract customer_id from the file path
2. Write transformed data to the Silver schema

## 1. Extract customer_id from the file path

```python
df = spark.table("gizmobox.bronze.py_memberships")
display(df)
```

```python
from pyspark.sql import functions as F

df_extracted_memberships = (
    df.select(
        F.regexp_extract(F.col("path"), r".*/([0-9]+)\.png$", 1).alias("customer_id"),
        F.col("content").alias("membership_card")
    )
)

display(df_extracted_memberships)
```

## 2. Write transformed data to the Silver schema

```python
df_extracted_memberships.writeTo("gizmobox.silver.py_memberships").createOrReplace()
```

```python
df = spark.table("gizmobox.silver.py_memberships")

display(df)
```
