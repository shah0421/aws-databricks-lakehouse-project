# Transform Orders Data - Explode Arrays
1. Access elements from the JSON object
2. Deduplicate Array Elements
3. Explode Arrays
4. Write the Transformed Data to Silver Schema

```python
df = spark.table("gizmobox.silver.orders_json")
display(df)
```

## 1. Access elements from the JSON object

```python
df_orders_normalized = (
    df.select(
        "json_value.order_id",
        "json_value.order_status",
        "json_value.payment_method",
        "json_value.total_amount",
        "json_value.transaction_timestamp",
        "json_value.customer_id",
        "json_value.items"
    )
)
display(df_orders_normalized)
```

## 2. Deduplicate Array Elements

```python
from pyspark.sql import functions as F

df_orders_normalized = (
    df.select(
        "json_value.order_id",
        "json_value.order_status",
        "json_value.payment_method",
        "json_value.total_amount",
        "json_value.transaction_timestamp",
        "json_value.customer_id",
        F.array_distinct("json_value.items").alias("items")
    )
)
display(df_orders_normalized)
```

## 3. Explode Arrays

```python
df_orders_exploded = (
    df_orders_normalized.select(
        "order_id",
        "order_status",
        "payment_method",
        "total_amount",
        "transaction_timestamp",
        "customer_id",
        F.explode("items").alias("item")
    )
)
display(df_orders_exploded)
```

## 4. Write the Transformed Data to Silver Schema

```python
df_order_items = (
    df_orders_exploded.select(
        "order_id",
        "order_status",
        "payment_method",
        "total_amount",
        "transaction_timestamp",
        "customer_id",
        "item.item_id",
        "item.name",
        "item.price",
        "item.quantity",
        "item.category",
        "item.details.brand",
        "item.details.color"
    )
)
display(df_order_items)
```

```python
df_order_items.writeTo("gizmobox.silver.py_orders").createOrReplace()
```

```python
df = spark.table("gizmobox.silver.py_orders")
display(df)
```
