3. Create Payments Table in Bronze Schema

## 1. List the files from Payment folder

```python
%fs ls 's3a://amzn-s3-gizmobox-project-bucket/gizmobox/landing/external-data/Payments/'
```

## 2. Read Payment File

```python
payment_schema = 'payment_id INTEGER, order_id INTEGER, payment_timestamp TIMESTAMP, payment_status INTEGER, payment_method STRING'
```

```python
from pyspark.sql.types import StructType, StructField, StringType, IntegerType, FloatType, TimestampType
py_payment_schema = StructType([
    StructField("payment_id", IntegerType()),
    StructField("order_id", IntegerType()),
    StructField("payment_timestamp", TimestampType()),
    StructField("payment_status", IntegerType()),
    StructField("payment_method", StringType())
])
```

```python
df = (
  spark.read.format("csv")
    .option("delimiter", ",")
    .schema(py_payment_schema)
    .load("s3a://amzn-s3-gizmobox-project-bucket/gizmobox/landing/external-data/Payments/")
)
display(df)
```

## 3. Create Payments Table in Bronze Schema

```python
df.writeTo("gizmobox.bronze.py_payments").createOrReplace() 
```

```python
data = spark.sql("Select * from gizmobox.bronze.py_payments;")
display(data)
```
